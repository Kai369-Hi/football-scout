# 01 · 下载 Understat 数据（2025-26 五大联赛）

**目的**：取得进攻 / 组织指标（npxG、xA、shots、key_passes、xGChain、xGBuildup），存进 `data/raw/`。

| 步骤 | 内容 |
|---|---|
| 1 | 设定路径 |
| 2 | 逐个联赛下载（一个联赛失败不影响其他） |
| 3 | 合并、检查、保存 |

💡 按 `Shift + Enter` 一格一格运行。第一次下载需要一点时间，之后 soccerdata 会**缓存**，第二次运行会快很多。

```python
from pathlib import Path
import pandas as pd
import soccerdata as sd

# 自动找到项目根目录（有 requirements.txt 的那个文件夹）
ROOT = Path.cwd()
while not (ROOT / "requirements.txt").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
RAW = ROOT / "data" / "raw"
print("项目根目录:", ROOT)
print("raw 文件夹存在:", RAW.exists())
print("soccerdata 版本:", sd.__version__)
```

## 2️⃣ 逐个联赛下载

用 `try / except` 包住每个联赛：🍜 就像五家供应商分开进货，一家缺货不会让整张订单取消。

```python
LEAGUES = ["ENG-Premier League", "ESP-La Liga", "GER-Bundesliga", "ITA-Serie A", "FRA-Ligue 1"]
SEASON = "2025/2026"

frames, failed = [], {}
for lg in LEAGUES:
    try:
        us = sd.Understat(leagues=lg, seasons=SEASON)
        df = us.read_player_season_stats().reset_index()
        frames.append(df)
        print(f"✅ {lg:20} {len(df):>4} 行")
    except Exception as e:
        failed[lg] = repr(e)
        print(f"❌ {lg:20} 失败: {e!r}")

print("\n失败的联赛:", list(failed) or "无")
```

## 3️⃣ 合并、检查、保存

```python
understat = pd.concat(frames, ignore_index=True)

print("形状 (行, 列):", understat.shape)
print("\n列名:", understat.columns.tolist())
print("\n各联赛行数:")
print(understat["league"].value_counts())
print("\n缺失值:")
print(understat.isna().sum()[understat.isna().sum() > 0])
display(understat.head())
```

```python
out = RAW / "understat_2025_26.csv"
understat.to_csv(out, index=False, encoding="utf-8")
print("✅ 已保存:", out)
```
